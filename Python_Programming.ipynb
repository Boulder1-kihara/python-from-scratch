# **Concept of Object Reference**
Let's assign a varaible `x` to value `5`.
`x = 5`

When `x = 5` is executed, python creates an object to represent the value `5` and makes `x` reference this object.

````python
x = 5
````

Now asign another variable `y` to the variable `x`.

`y = x`

This statement creates `y` and references the same object as `x`, not `x` itself. This is called a **Shared Reference**

A **shared reference** is where multiple variables reference the same object.

````python
x = 5
y = x

print(x, y)
````

image.png

Now write `x = "Abel"`. Python creates a new obkect for the value `Abel`and makes `x` reference this new object

````python
x = 5
y = x
x = "Abel"

print(x, y)
````

The variable `y` remains unchanged, still referencing the original object `5`. Now, If we assign a new value to `y`

`y = computer`

Now `5` will be trash

````python
x = 5
y = x
x = "Abel"
y = "computer"

print(x, y)
````

# **Deleting a Variable**
`del` keyword is used to delete a variable from memory. After deletion, the variable can no longer be accessed producing an error.

````python
x = 5
y = "abel"
del y
print(x, y)
````

In Python, using `del` on standalone variables is rarely necessary because Python's garbage collector automatically frees up memory when variables are no longer used. To prevent the error from occurring, asign `None`.

````python
x = 5
y = "abel"

y = None  # Clears the value while keeping 'y' defined

print(x, y)  # Outputs: 5 None
````

Or use `try...except` block to gracefully handle the missing variable without crashing your program.

````python
x = 5
y = "abel"
del y

try:
    print(x, y)
except NameError:
    print(x, "y is not defined")
````

<u>**Arithmetic Operators**</u>

These are used to perform basic mathematical operations like additon, subtraction, division and multiplication.

In python, the division operator (/)returns a floating point result, while floor division(//), returns a whole integer result.

````python
from binascii import b2a_base64
a = 5
b = 3

#sum
sum = a + b
print(f"the sum of a and b is {sum}")

#subtraction
subtraction = a - b
subtractions = b - a
print(f"the deviation of a and b is {subtraction} and {subtractions}")

#multiplication
product = a*b
print(f"the product of a and b is {product}")

#division
division = a/b
print(f"the division of a and b is {division}")

#floor division
divisionb = a//b
print(f"the floor division of a and b is {divisionb}")

#modulus(displays the remainder)
modulus = a%b
print(f"the modulus of a and b is {modulus}")

#exponentiation
exponentiation = a**b
print(f"the exponentiation of a and b is {exponentiation}")
````

<u>**Comparison Operator**<u/>

````python
a = 5
b = 3

#returns true
print(a > b)

#returns false
print(a < b)

#returns false
print(a == b)

#returns true
print(a >= b)

#returns false
print(a <= b)

#returns true
print(a != b)
````

## **If-elif-else**
elif statement is used to check multiple conditions in a program.

it executes a block of code when its condition evaluates to true after previous conditions evaluate to false.

````python
age = int(input("enter your age: "))

if age <= 12:
  print("you are a child")

elif age <= 19:
  print("you are a teenager")

elif age <= 35:
  print("you are a young adult")

else:
  print("adult")
````

## **Nested if else statement**

These are if-else statements placed inside another if or else block.

It is used to check conditions within another condition.

````python
cost = 2000
is_member = True

if cost >= 1500:
    if is_member:
      print("20% discount")

    else:
      print("10% discount")

else:
  print("you are not eligible for any discount")
````

## **Match case statement**

A match case statement is used to compare a value against multiple patterns and execute the matching block of code.

It is similar to the switch case statement available in other programming languages.

````python
number = 3

match number:
  case 1:
    print("one")

  case 2|3:
    print("two or three")

  case _:
    print("other number")
````
