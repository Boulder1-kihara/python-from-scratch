<u>**Printing Multiple Variables**</u>

To print multiple variables, we seperate the variables using commas in the print function.



```

```



````python
name = "Abel"
age = "20"
print(name, age, "years")
````

We can also accept multiple inputs at once and ise the `.split()` method making them seperate




````python
x, y = input("enter 2 numbers: ").split()
print(x, y)
````

# **Rules for Naming Variables**

1. Names can contain letters, digits and underscores (_).

    e.g

    ```
    first_name
    ```



2. The first character cannot be a digit.

    e.g

        correct:
        num1

        wrong:
        1num

3. Names are case-sensitive, so myVar and myvar are treated differently.


        correct:
        myVar = "num1"
        num1 = 3

        wrong:
        1myvar = "num1"
        num1 = 3


4. Keywords such as if, else and for cannot be used as variable names.
